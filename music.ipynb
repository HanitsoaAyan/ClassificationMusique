```python
import pandas as pd
import numpy as np
```

```python
music = pd.read_csv('features_30_sec.csv')
music.shape
```

```python
music.describe()
```

```python
music = music.drop(columns=['length'])
```

```python
colonnes = [c for c in music.columns if c not in ('filename','label')]
print(len(colonnes))
```

```python
music.head(10)
```

```python
print(music.duplicated(subset=colonnes).sum())
```

```python
masque = music.duplicated(subset=colonnes, keep=False)
print(music[masque].sort_values(colonnes[0])[['filename', 'label']])

```

```python
music = music[~music['filename'].isin(['rock.00016.wav', 'metal.00058.wav'])]
```

```python
music = music.drop_duplicates(subset=colonnes, keep='first')
```

```python
music = music.reset_index(drop=True)
```

```python
print(music.duplicated(subset=colonnes).sum())
```

```python
music.shape
```

```python
import numpy as np

# Découpage stratifié 80 % apprentissage / 20 % test, graine fixe
rng = np.random.default_rng(42)
idx_train, idx_test = [], []

for genre, groupe in music.groupby('label'):
    idx = rng.permutation(groupe.index.to_numpy())
    n_test = int(round(len(idx) * 0.2))
    idx_test.extend(idx[:n_test])   # 20 % pour le test
    idx_train.extend(idx[n_test:])  # 80 % pour l'apprentissage

train = music.loc[idx_train]
test = music.loc[idx_test]

print(train.shape, test.shape)  # (787, 59) (198, 59)
```

```python
print("Total :", len(train) + len(test))  # 985
print(test['label'].value_counts().sort_index())
print((test['label'].value_counts() / music['label'].value_counts()).sort_index())
```

```python
import numpy as np

# 1. Normalisation : moyenne et écart-type calculés sur le TRAIN uniquement
moyenne = train[colonnes].mean()
ecart = train[colonnes].std()
ecart[ecart == 0] = 1  # sécurité : caractéristique constante

X_train = ((train[colonnes] - moyenne) / ecart).to_numpy()
X_test = ((test[colonnes] - moyenne) / ecart).to_numpy()

# 2. Encodage des genres en 0..9 (ordre alphabétique)
genres = sorted(music['label'].unique())
code = {g: i for i, g in enumerate(genres)}
y_train = train['label'].map(code).to_numpy()
y_test = test['label'].map(code).to_numpy()

# 3. Vérifications
print("X_train :", X_train.shape)  # (787, 57)
print("X_test  :", X_test.shape)   # (198, 57)
print("y_train :", y_train.shape, "| y_test :", y_test.shape)
print("Genres  :", genres)
print("Moyenne X_train (≈ 0) :", X_train.mean())
print("Écart-type X_train (≈ 1) :", X_train.std())
print("NaN dans X_train :", np.isnan(X_train).sum(), "| dans X_test :", np.isnan(X_test).sum())
```

```python
import os
os.makedirs('data', exist_ok=True)

# La table nettoyée (985 morceaux) : dans ce notebook elle s'appelle music
music.to_csv('data/music_propre.csv', index=False)

# Les tableaux prêts pour B et C
np.save('data/X_train.npy', X_train)
np.save('data/X_test.npy', X_test)
np.save('data/y_train.npy', y_train)
np.save('data/y_test.npy', y_test)
```

```python
import numpy as np

def precision(y_vrai, y_pred):
    """Pourcentage de bonnes réponses."""
    return float(np.mean(np.asarray(y_vrai) == np.asarray(y_pred)))

def matrice_confusion(y_vrai, y_pred, n_classes=10):
    """Lignes = vraie classe, colonnes = classe prédite."""
    mat = np.zeros((n_classes, n_classes), dtype=int)
    for vrai, pred in zip(y_vrai, y_pred):
        mat[int(vrai), int(pred)] += 1
    return mat

# Test immédiat avec des prédictions factices (toujours blues = 0)
y_faux = np.zeros_like(y_test)
print("Précision (factice) :", precision(y_test, y_faux))
print(matrice_confusion(y_test, y_faux))
```
